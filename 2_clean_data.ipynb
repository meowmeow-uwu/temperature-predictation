```python
import pandas as pd
import os
```

```python
df_train = pd.read_csv('raw_data/raw_data_train.csv')
df_train.head()
```

```python
df_test = pd.read_csv('raw_data/raw_data_test.csv')
df_test.head()
```

```python
print(df_train.dtypes, df_test.dtypes, sep='\n')
```

```python
output_dir = "clean_data"
os.makedirs(output_dir, exist_ok=True)

# Chuyển đổi cột thời gian
df_train['date'] = pd.to_datetime(df_train['date']).dt.tz_localize(None)
df_test['date'] = pd.to_datetime(df_test['date']).dt.tz_localize(None)
df_train = df_train.rename(columns={'date': 'time'})
df_test = df_test.rename(columns={'date': 'time'})

#Xóa trùng lặp
df_train = df_train.drop_duplicates(subset=['time'])
df_test = df_test.drop_duplicates(subset=['time'])

# Sắp xếp lại thời gian
df_train = df_train.sort_values(by='time').reset_index(drop=True)
df_test = df_test.sort_values(by='time').reset_index(drop=True)

# Đưa cột 'time' làm Index để ép tần suất thời gian liên tục từng giờ (.asfreq('H'))
# Nếu có giờ nào bị khuyết dữ liệu, Pandas sẽ tự động chèn thêm 1 dòng trống (NaN)
# Cuối cùng, dùng .reset_index() để trả cột 'time' về làm một cột bình thường 
df_train = df_train.set_index('time').asfreq('h').reset_index()
df_test = df_test.set_index('time').asfreq('h').reset_index()

train_path = os.path.join(output_dir, "clean_data_train.csv")
test_path = os.path.join(output_dir, "clean_data_test.csv")

df_train.to_csv(train_path, index=False, encoding="utf-8-sig")
df_test.to_csv(test_path, index=False, encoding="utf-8-sig")

print(f" - Train: {len(df_train)} dòng (Lưu tại {train_path})")
print(f" - Test:  {len(df_test)} dòng (Lưu tại {test_path})")
```
